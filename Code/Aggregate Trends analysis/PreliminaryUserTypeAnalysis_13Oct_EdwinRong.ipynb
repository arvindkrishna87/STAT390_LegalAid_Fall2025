```python
import pandas as pd
import numpy as np

# Adjusting how much pandas shows when displaying dfs/series
pd.set_option('display.max_rows', 1000)
pd.set_option('display.max_columns', 100)
```

This Jupyter notebook contains some processing for analysis and the output of the distribution of User type at the last steps of all customer calls (for a preliminary visualization).

This contains only the August 2025 All Calls data as processing the entire dataset takes a very long time (over 30 minutes).

# Cleaning AllCalls

```python
# Read dataset
allcalls = pd.read_csv('Data/Data/All Calls by Month/August 2025.csv')
```

```python
# Get rid of columns that are full of NaN
allcalls_no_nan = allcalls.dropna(axis=1, how='all')
allcalls_no_nan.head()
```

```python
# Convert columns with mixed data types to only string
mixed_dtype_cols = ['Model', 'International Country', 'Redirecting number',
                    'PSTN vendor name', 'Device owner UUID', 'Redirecting party UUID',
                    'PSTN vendor name2', 'User']

# The following have NaN values that make them mixed when they should be all string (NaN values are float)
mixed_dtype_cols = mixed_dtype_cols + ['Redirect reason', 'Local SessionID',
       'Final local sessionID', 'PSTN provider ID', 'Remote SessionID',
       'Answer time', 'Remote call ID', 'Call transfer time',
       'Sub client type', 'Device Mac', 'User UUID', 'Final remote sessionID',
       'Original reason', 'Inbound trunk', 'Network call ID', 'Outbound trunk',
       'Transfer related call ID', 'Call ID', 'Client version',
       'Related call ID', 'Related reason', 'Call outcome reason',
       'External caller ID number']

allcalls_no_mixed = allcalls_no_nan.copy()

for col in mixed_dtype_cols:
    if col in allcalls_no_mixed.columns:
        allcalls_no_mixed[col] = allcalls_no_mixed[col].astype(str)
```

```python
# Turn columns with phone numbers into strings
for col in ['Called number', 'Redirecting number', 'User number', 'Site main number', 'External caller ID number']:
    allcalls_no_mixed[col] = allcalls_no_mixed[col].astype(str)
    # For column(s) with phone numbers represented as float
    allcalls_no_mixed[col] = allcalls_no_mixed[col].apply(lambda x: x[:-2] if x[-2:] == '.0' else x)
```

# Filtering

```python
# Split data into unique calls by Correlation ID
grouped = allcalls_no_mixed.groupby('Correlation ID')
```

```python
# Get only calls that are from customers (Inbound trunk = NA, Outbound trunk = wcc_..., Direction = TERMINATING, Call type = SIP_INBOUND, Client type = WXCC)
customer_calls = dict()

for name, df in grouped:
    first_leg = df.iloc[-1,:]
    if first_leg['Outbound trunk'] != 'nan' and first_leg['Direction'] == 'TERMINATING' and first_leg['Call type'] == 'SIP_INBOUND' and first_leg['Client type'] == 'WXCC':
        customer_calls[name] = df
```

```python
# For each possible User type, get number of calls (Correlation IDs) that end in a row with that User type
user_type_ending_distribution = dict() # This is a dict of counts per user type
user_type_ending_collection = dict() # This dict of dicts will contain the correlation id & actual df for each call within each user type 

for user_type in allcalls_no_mixed['User type'].unique():
    user_type_ending_distribution[user_type] = 0
    user_type_ending_collection[user_type] = dict()

for corr_id in customer_calls.keys():
    call = customer_calls[corr_id]
    last_leg = call.iloc[0,:]
    user_type_ending = last_leg['User type']
    user_type_ending_distribution[user_type_ending] += 1

    user_type_ending_collection[user_type_ending][corr_id] = call
```

The User type column is useful as it can indicate where customer calls are ended. For example, if the last row of a call has the User type == User, that means the call ended with a live agent on the line.

```python
# Export to a csv file that Power BI can use
df = pd.DataFrame(pd.Series(user_type_ending_distribution))
df.name = 'End user type counts'
df.reset_index(inplace=True)
df.columns = ['User type', 'Count']

df.to_csv('End_user_type_counts.csv', index=False)
```
