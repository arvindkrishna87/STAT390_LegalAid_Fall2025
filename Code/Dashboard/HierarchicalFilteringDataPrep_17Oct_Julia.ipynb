It also groups specific single-level menus under a new Tier1 category called "Miscellaneous" to make filters easier to sort through. The resulting CSV file should be uploaded into Power BI and linked with your CAR data files via the "Activity Name" column to enable hierarchical filtering (when you do this, make sure the cross-filtering setting is set to "Both").
```python
import pandas as pd
import numpy as np

combined_calls = pd.read_csv("combined_calls.csv")

# --- Step 1: Your raw activity list ---
activity_list = [
    "FarmworkerMainMenu 1",
    "MainMenu 1",
    "AddressFaxHoursMenu 2",
    "ComplimentOrComplaintMenu 2",
    "AppointmentMenu 2",
    "HolidayPrompt 2",
    "LanguageSelectionMenu 2",
    "StaffDirectoryEnglishTransfer 2",
    "StaffDirectorySpanishTransfer 2",
    "FrontDeskTransfer 1",
    "FrontDeskTransfer1 2",
    "FrontDeskTransfer2 2",
    "FrontDeskTransfer3 2",
    "ClosedQueueMenu 1",
    "ClinicVoicemailTransfer 2",
    "SeniorsMenu 1",
    "SeniorsConfirmationMenu 2",
    "SuburbsOrCityMenu 2",
    "SeniorNotCookCoMenu 3",
    "ThankYouGoodbye 4",
    "SuburbanSeniorsMenu 3",
    "SubSeniorHomeownerQueue 4",
    "GetLoggedInSubSeniorHomeownerSPAgents 4",
    "GetLoggedInSubSeniorHomeownerAgents 4",
    "GetLoggedInSubSeniorBenefitsSPAgents 4",
    "SubSeniorHomeownerSPQueue 4",
    "GetLoggedInSubSeniorOtherAgents 4",
    "GetLoggedInSubSeniorTenantAgents 4",
    "SubSeniorOtherQueue 4",
    "SubSeniorPreQueueMessage1_1 4",
    "GetLoggedInSubSeniorConsumerAgents 4",
    "SubSeniorConsumerQueue 4",
    "SubSeniorTenantQueue 4",
    "GetLoggedInSubSeniorEmploymentSPAgents 4",
    "GetLoggedInSubSeniorBenefitsAgents 4",
    "GetLoggedInBenefitsSubSeniorsAgents 4",
    "SubSeniorPreQueueMessage1_2 4",
    "BenefitsSubSeniorsQueue 4",
    "GetLoggedInSubSeniorFamilyAgents 4",
    "SubSeniorFamilyQueue 4",
    "SubSeniorBenefitsQueue 4",
    "GetLoggedInSubSeniorTenantSPAgents 4",
    "SubSeniorTenantSPQueue 4",
    "GetLoggedInSubSeniorEmploymentAgents 4",
    "SubSeniorEmploymentQueue 4",
    "GetLoggedInSubSeniorOtherSPAgents 4",
    "GetLoggedInSubSeniorFamilySPAgents 4",
    "GetLoggedInFamilySubSeniorsAgents 4",
    "GetLoggedInSubSeniorConsumerSPAgents 4",
    "SubSeniorConsumerSPQueue 4",
    "GetLoggedInEmploymentSPAgents 4",
    "GetLoggedInHousingSubSeniorsAgents 4",
    "GetLoggedInConsumerSubSeniorsAgents 4",
    "GetLoggedInEmploymentSubSeniorsAgents 4",
    "GetLoggedInHomeownerSubSeniorsAgents 4",
    "GetLoggedInOtherSubSeniorsSPAgents 4",
    "GetLoggedInBenefitsSubSeniorsSPAgents 4",
    "GetLoggedInHousingSubSeniorsSPAgents 4",
    "GetLoggedInFamilySubSeniorsSPAgents 4",
    "GetLoggedInConsumerSubSeniorsSPAgents 4",
    "GetLoggedInEmploymentSubSeniorsSPAgents 4",
    "HousingSubSeniorsQueue 4",
    "EmploymentSubSeniorsSPQueue 4",
    "OtherSubSeniorsQueue 4",
    "FamilySubSeniorsSPQueue 4",
    "HousingSubSeniorsSPQueue 4",
    "BenefitsSubSeniorsSPQueue 4",
    "ConsumerSubSeniorsQueue 4",
    "EmploymentSubSeniorsQueue 4",
    "OtherSubSeniorsSPQueue 4",
    "FamilySubSeniorsQueue 4",
    "HomeownerSubSeniorsQueue 4",
    "ConsumerSubSeniorsSPQueue 4",
    "SubSeniorEmploymentSPQueue 4",
    "GetLoggedInOtherSubSeniorsAgents 4",
    "SubSeniorOtherSPQueue 4",
    "SubSeniorFamilySPQueue 4",
    "SubSeniorBenefitsSPQueue 4",
    "SeniorsADAPTMenu 4",
    "GetLoggedInSubSeniorADAPTAgents 5",
    "SubSeniorADAPTQueue 5",
    "GetLoggedInSubSeniorADAPTSPAgents 5",
    "GetLoggedInADAPTSubSeniorsAgents 5",
    "ADAPTSubSeniorsQueue 5",
    "SubSeniorADAPTSPQueue 5",
    "GetLoggedInADAPTSubSeniorsSPAgents 5",
    "ADAPTSubSeniorsSPQueue 5",
    "LegalMenu1 1",
    "LegalMenu2 2",
    "ADAPTQueue 3",
    "ADAPTSPQueue 3",
    "GetLoggedInADAPTSPAgents 3",
    "GetLoggedInADAPTAgents 3",
    "CriminalRecordsVoicemailTransfer 3",
    "ConsumerSPQueue 3",
    "ConsumerQueue 3",
    "GetLoggedInConsumerSPAgents 3",
    "GetLoggedInConsumerAgents 3",
    "HousingMenu 3",
    "PreTenantMenu 4",
    "TenantMenu 5",
    "TenantDeterrenceMenu 5",
    "GetLoggedInHousingAgents 4",
    "HousingQueue 4",
    "GetLoggedInHousingSPAgents 4",
    "HousingSPQueue 4",
    "EmploymentMenu 3",
    "GetLoggedInEmploymentAgents 4",
    "EmploymentQueue 4",
    "EmploymentSPQueue 4",
    "WorkersCompMenu 4",
    "ImmigrationMenu 3",
    "GetLoggedInImmigrationSPAgents 4",
    "ImmigrationOtherMenu 4",
    "ImmigrationSPQueue 4",
    "GetLoggedInImmigrationAgents 4",
    "ImmigrationQueue 4",
    "MigrantVoicemailTransfer 4",
    "TraffickingVoicemailTransfer 4",
    "FamilyMenu 3",
    "DivorceOrParentingMenu 4",
    "SimpleDivorceMenu 5",
    "ChildSupportMenu 5",
    "FamilyQueue 4",
    "EducationQueue 4",
    "EducationSPQueue 4",
    "GetLoggedInEducationAgents 4",
    "GetLoggedInEducationSPAgents 4",
    "GetLoggedInFamilySPAgents 4",
    "GetLoggedInFamilyAgents 4",
    "FamilySPQueue 4",
    "OtherLegalMenu 3",
    "OtherLegalCriminalCaseMenu 4",
    "OtherLegalPersonalInjuryMenu 4",
    "OtherLegalOtherMenu 4",
    "BenefitsMenu 3",
    "BenefitsQueue 4",
    "BenefitsSPQueue 4",
    "GetLoggedInBenefitsAgents 4",
    "VeteransBenefitsVoicemailTransfer 4",
    "HIVMenu 3",
    "HIVVoicemailTransfer 4",
    "DisconnectContact 1",
    "DisconnectContact1 2",
    "DisconnectContact2 2",
    "DisconnectCallbackContact 2",
    "CCB 1",
    "PlayCCBConfirmation 2",
    "IntakePreQueueMessage1 1",
    "PreQueueMessage2 2",
    "LegalServerScreenPop 2",
    "ScreenPopProcessComplete 2",
    "PlayMOH300s 2",
    "QueueMenu1 2",
    "SetCallerID 1",
    "HelpWithLegalorOtherReasonMenu 1",
    "ClosedMenu 1",
    "ReadANI 1",
    "CallbackRetry 1",
    "ConfirmCallbackNumber 1",
    "CollectCallbackNumber 1",
    "PlayErrorMessage 1",
    "TransferToSafeHaven 1"
]

# --- Step 2: Extract Activity Name and Tier ---
rows = []
for line in activity_list:
    line = line.strip()
    if not line:
        continue
    if " " in line:
        name, tier_str = line.rsplit(" ", 1)
        tier = int(tier_str)
        rows.append({"Activity_Name": name, "Tier": tier})

df = pd.DataFrame(rows)

# --- Step 3: Adjust tiers for "Miscellaneous" group ---
misc_group = [
    "SetCallerID",
    "HelpWithLegalorOtherReasonMenu",
    "ClosedMenu",
    "ReadANI",
    "CallbackRetry",
    "ConfirmCallbackNumber",
    "CollectCallbackNumber",
    "PlayErrorMessage",
    "TransferToSafeHaven"
]

df.loc[df["Activity_Name"].isin(misc_group), "Tier"] = 2
df["Tier1"] = np.nan
df["Tier2"] = np.nan
df["Tier3"] = np.nan
df["Tier4"] = np.nan
df["Tier5"] = np.nan

# --- Step 4: Build hierarchy ---
last_seen = {}

for idx, row in df.iterrows():
    tier = row["Tier"]

    # If this is a "Miscellaneous" item, set Tier1 explicitly
    if row["Activity_Name"] in misc_group:
        df.at[idx, "Tier1"] = "Miscellaneous"
    else:
        last_seen[tier] = row["Activity_Name"]

    # Fill hierarchy
    for i in range(1, 6):
        if i <= tier:
            if row["Activity_Name"] in misc_group and i == 1:
                continue  # already set Tier1 = Miscellaneous
            df.at[idx, f"Tier{i}"] = last_seen.get(i, np.nan)
        else:
            df.at[idx, f"Tier{i}"] = np.nan

# --- Step 5: Replace any None with NaN for clean Power BI filtering ---
df = df.replace({None: np.nan})

# --- Step 6: Check result ---
print(df.head(20))
```
