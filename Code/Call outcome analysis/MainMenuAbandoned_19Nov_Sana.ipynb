```python
# Create sequential call_id correctly

# 1) Sort df_main properly (and assign)
df_main = df_main.sort_values(
    by=['Contact Session ID', 'Activity Start Timestamp']
).reset_index(drop=True)

# 2) Create sequential call_id (one id per Contact Session ID)
# Option A
df_main['call_id'] = (
    (df_main['Contact Session ID'] != df_main['Contact Session ID'].shift())
    .cumsum()
)
# REORDER COLUMNS
column_order = [
    'call_id',
    'Contact Session ID',
    'EP Name',
    'Flow Name',
    'Activity Name',
    'Activity Start Timestamp',
    'Queue Name',
    'Agent Name',
    'Termination Reason']
df_main = df_main[column_order]

df_main.head()
```

```python
# Make sure rows are sorted by call/session and timestamp
df_main = df_main.sort_values(['call_id', 'Activity Start Timestamp']).reset_index(drop=True)

# next termination reason if you need it
df_main['next_termination_reason'] = (
    df_main.groupby('call_id')['Termination Reason'].shift(-1)
)
```

```python
# Step 1: Get all unique Contact Session IDs that went through the main MENU
main_call_ids = main_ep_calls['Contact Session ID'].unique()

# Step 2: Filter df_main to only include those same sessions
df_main_sessions = df_main[df_main['Contact Session ID'].isin(main_call_ids)].copy()

# Step 3: sort chronologically to see the call flow order
df_main_sessions = df_main_sessions.sort_values(
    by=['Contact Session ID', 'Activity Start Timestamp']
).reset_index(drop=True)
```

```python
# For each call_id, take the last non-null Activity Name
df_main_sessions = df_main_sessions.sort_values(
    ['call_id', 'Activity Start Timestamp']
)

df_main_sessions['Last_Activity'] = (
    df_main_sessions
      .groupby('call_id')['Activity Name']
      .transform(lambda s: s.dropna().iloc[-1] if s.notna().any() else np.nan)
)
```

```python
# Identify the last EP Name per Contact Session ID
last_ep = (
    df_main_sessions.groupby('Contact Session ID')['EP Name']
    .last()
    .reset_index()
    .rename(columns={'EP Name': 'Last_EP_Name'})
)

# Merge that back to the main df_benefits
df_main_sessions = df_main_sessions.merge(last_ep, on='Contact Session ID', how='left')

# Create Yes/No or 1/0 indicator
df_main_sessions['ended_in_main_ep'] = np.where(
    df_main_sessions['Last_EP_Name'] == 'Main Number Telephony EP', 'Yes', 'No'
)
```

```python
## activities that are a transfer.. finding if an agent connected or not. 
# This runs on the logic that if an agent hadn't connected, the call is abandoned.
# Also removing calls that were transferred after the AppointmentMenu immediately, again, AppointmentMenu is not in Main Menu. 
# Alternatively, all these transfers can be considered NOT ABANDONED because they would have reached a transfer, 
# customer might have just chosen not to! 
#################


# STEP 1: sort and build previous activity
df_filtered = df_filtered.sort_values(
    ['call_id', 'Activity Start Timestamp']
).copy()
df_filtered['prev_activity'] = df_filtered.groupby('call_id')['Activity Name'].shift(1)

# STEP 2: find call_ids to exclude (AppointmentMenu --> transfer)
activities = [
    'FrontDeskTransfer1',
    'FrontDeskTransfer2',
    'FrontDeskTransfer3',
    'StaffDirectoryEnglishTransfer',
    'StaffDirectorySpanishTransfer',
    'ClinicVoicemailTransfer'
]

bad_call_ids = df_filtered[
    (df_filtered['Activity Name'].isin(activities)) &
    (df_filtered['prev_activity'] == 'AppointmentMenu')
]['call_id'].unique()

# STEP 3: remove those call_ids
df_filtered2 = df_filtered[~df_filtered['call_id'].isin(bad_call_ids)]

# STEP 4: analyses
rows = []
ids_by_activity = {}

for act in activities:
    df_act = df_filtered2[df_filtered2['Last_Activity'] == act]

    per_call = (
        df_act
        .groupby('call_id')['Agent Name']
        .apply(lambda s: s.notna().any())
        .rename('agent_connected')
        .reset_index()
    )

    connected_count = per_call['agent_connected'].sum()
    not_connected_count = (~per_call['agent_connected']).sum()
    total_unique_calls = per_call.shape[0]

    rows.append({
        'Activity': act,
        'Agent Connected (unique calls)': int(connected_count),
        'Agent NOT Connected (unique calls)': int(not_connected_count),
        'Total unique calls': int(total_unique_calls),
        'Pct connected': connected_count / total_unique_calls if total_unique_calls else np.nan
    })
    # store the call_id lists 
    ids_by_activity[act] = {
        'connected_call_ids': per_call.loc[per_call['agent_connected'], 'call_id'].to_numpy(),
        'not_connected_call_ids': per_call.loc[~per_call['agent_connected'], 'call_id'].to_numpy()
    }

out = pd.DataFrame(rows)
out['Pct connected'] = (out['Pct connected'] * 100).round(1)
print(out.to_string(index=False))
```

```python
import matplotlib.pyplot as plt

# --- Data ---
activities = [
    'Front Desk Transfer 2',
    'Front Desk Transfer 3',
    'Clinic Voicemail Transfer',
    'Staff Directory English Transfer',
    'Staff Directory Spanish Transfer',
    'Help with Legal or Other',
    'Main Menu'
]

counts = [0, 97, 3, 2, 1, 34, 13072]
total_calls = 165680
percents = [round(c / total_calls * 100, 2) for c in counts]

# Sort ascending
sorted_pairs = sorted(zip(counts, activities, percents))
sorted_counts, sorted_activities, sorted_percents = zip(*sorted_pairs)

plt.figure(figsize=(12,6))
bars = plt.barh(sorted_activities, sorted_counts, color='skyblue')

# Annotate
for i, (bar, c, p) in enumerate(zip(bars, sorted_counts, sorted_percents)):
    x = bar.get_width()
    y = bar.get_y() + bar.get_height()/2
    label = f"{c} ({p}%)"
    
    # For the largest bar (Main Menu), put inside
    if c == max(counts):
        plt.text(x - 500, y, label, va='center', ha='right', color='black', fontsize=12)
    else:
        plt.text(x + 50, y, label, va='center', ha='left', fontsize=12)

plt.xlabel('Number of Calls', fontsize=12)
plt.title('Call Abandonment with All Calls Entering the Main Menu:', fontsize=14)
plt.tight_layout()
plt.show()
```
