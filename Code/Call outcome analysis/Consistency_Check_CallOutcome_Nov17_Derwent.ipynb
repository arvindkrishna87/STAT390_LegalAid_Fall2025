### Reading all filenames in the data folder

```python
folder = Path(data_path)
files = sorted(list(folder.glob("*.csv")) + list(folder.glob("*.xlsx")))
df_main = pd.DataFrame(columns=['Contact Session ID', 'EP Name', 'Flow Name', 'Activity Name', 'Activity Start Timestamp', 
                                'Queue Name', 'Agent Name', 'Termination Reason'])
df_main
```

### Reading all data files
The code chunk below reads and appends all the CAR data files. The first two rows of each file are blank and thus ignored.

```python
i=0
for f in files:
    i = i + 1
    if f.suffix.lower() == ".csv":
        df = pd.read_csv(f, header=2, dtype=str, engine="python", skip_blank_lines=False)
    else:  # .xlsx
        df = pd.read_excel(f, sheet_name=0, header=2, dtype=str)
    df_main = pd.concat([df_main, df], ignore_index=True)
    print(i, f.stem, df.shape)
```

```python
df_main.head()
```

### Time datatype conversion
The code chunk below converts time from string to datetime datatype.

```python
df_main["Activity Start Timestamp"] = df_main["Activity Start Timestamp"].apply(
    lambda x: datetime.strptime(x, "%Y/%m/%d %I:%M:%S %p"))
```

```python
# Checking the datatype of all columns
df_main.dtypes
```

```python
# Creating a new column 'hour' as it will be useful to visualize peak calling hours
df_main["hour"] = df_main["Activity Start Timestamp"].dt.hour
```

```python
# create date category in datetie
df_main["date"] = df_main["Activity Start Timestamp"].dt.date
df_main["date"] = pd.to_datetime(df_main["date"])
```

```python
# List of all last activities
last_act = [
    "LanguageSelectionMenu","ClinicVoicemailTransfer","MainMenu", "DisconnectContact1", "LegalMenu2", "FarmworkerMainMenu",
    "SetCallerID", "LegalServerScreenPop", "FrontDeskTransfer1", "ClosedMenu",
    "FrontDeskTransfer2", "OtherLegalMenu", "StaffDirectorySpanishTransfer",
    "FrontDeskTransfer", "TenantMenu", "CriminalRecordsVoicemailTransfer",
    "OtherLegalOtherMenu", "DivorceOrParentingMenu", "QueueMenu1", "LegalMenu1",
    "SeniorsMenu", "TenantDeterrenceMenu", "HousingMenu", "HIVVoicemailTransfer",
    "PlayMOH300s", "HIVMenu", "ScreenPopProcessComplete", "AddressFaxHoursMenu",
    "PreTenantMenu", "FamilyMenu", "FrontDeskTransfer3", "ChildSupportMenu",
    "ImmigrationMenu", "SeniorsADAPTMenu", "OtherLegalPersonalInjuryMenu",
    "OtherLegalCriminalCaseMenu", "SuburbsOrCityMenu", "SuburbanSeniorsMenu",
    "TransferToSafeHaven", "SimpleDivorceMenu", "BenefitsMenu",
    "SubSeniorPreQueueMessage1_1", "ThankYouGoodbye", "IntakePreQueueMessage1",
    "SeniorsConfirmationMenu", "VeteransBenefitsVoicemailTransfer",
    "EmploymentMenu", "No activity", "DisconnectContact2",
    "TraffickingVoicemailTransfer", "AppointmentMenu", "SeniorNotCookCoMenu",
    "ImmigrationOtherMenu", "HolidayPrompt", "DisconnectCallbackContact",
    "MigrantVoicemailTransfer", "ComplimentOrComplaintMenu", "PreQueueMessage2",
    "HelpWithLegalorOtherReasonMenu", "DisconnectContact", "WorkersCompMenu",
    "GetLoggedInSubSeniorTenantAgents", "ReadANI", "PlayErrorMessage",
    "CallbackRetry", "GetLoggedInSubSeniorOtherAgents", "CollectCallbackNumber",
    "ConfirmCallbackNumber", "SubSeniorPreQueueMessage1_2",
    "GetLoggedInSubSeniorOtherSPAgents", "SubSeniorOtherQueue",
    "GetLoggedInConsumerAgents", "ImmigrationSPQueue",
    "GetLoggedInOtherSubSeniorsAgents", "SubSeniorTenantQueue", "EducationSPQueue",
    "PlayCCBConfirmation", "SubSeniorHomeownerQueue", "SubSeniorConsumerQueue",
    "GetLoggedInSubSeniorTenantSPAgents"
]

# Store results for each activity
results = {}

for cat in last_act:
    # Find sessions whose LAST activity = this category
    ids = last[last["Last Activity"] == cat]["Contact Session ID"]

    # Extract full journeys for those sessions
    journeys = df_main[df_main["Contact Session ID"].isin(ids)]
    journeys = journeys.groupby('Contact Session ID').tail()

    # Save into dictionary
    results[cat] = {
        "ids": ids,
        "journeys": journeys
    }
  
```

```python
# verify the shape matches Krish's analysis
for cat in last_act:
    print(cat, ":", len(results[cat]["ids"]))
```

```python
# looping through every last activity to see how the calls ends generally (apologies for the long output)
for cat in last_act:
    grouped = results[cat]['journeys'].groupby("Contact Session ID")[['Activity Name', 'Termination Reason']]
    last_activities = grouped.apply(lambda x: x.tail(3))
    print(cat, ":\n", last_activities, "\n")
```
